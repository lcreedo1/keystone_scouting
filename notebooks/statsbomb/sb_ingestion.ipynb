```python
from statsbombpy import sb
import pandas as pd
from keystone_scouting.utilities.bq_utils import BigQueryConnector

GCP_PROJECT = "lgc-data"
GCP_DATASET = "keystone_scouting"
pipeline = BigQueryConnector(GCP_PROJECT)

%load_ext autoreload
%autoreload 2
```

```python
# get competitions and display all men's champions league rows
comps = sb.competitions()
display(comps[comps['competition_id'] == 16])
```

```python
# Get all matches from all seasons of men's champions league (competition_id = 16)
ucl_seasons = comps[comps['competition_id'] == 16]['season_id'].tolist()
ucl_matches = pd.concat([sb.matches(competition_id=16, season_id=season_id) for season_id in ucl_seasons], ignore_index=True)
display(ucl_matches)
```

```python
pipeline.upload_dataframe(ucl_matches, GCP_DATASET, "sb_matches")
```

```python
import time
import json
print("🚀 Commencing JSON-variant micro-batch stream...")

for i, match in enumerate(ucl_matches.iterrows(), 1):
    match_id = match[1]['match_id']
    print(f"📦 [{i}/{len(ucl_matches)}] Streaming Match ID: {match_id} to JSON block...")
    
    # 1. Fetch raw events (keep columns exactly as they come)
    events_df = sb.events(match_id=match_id)
    
    # 2. Build a clean, structured JSON payload for every single row
    json_records = events_df.to_dict(orient="records")
    
    # 3. Build a compact, predictable 3-column DataFrame
    staging_df = pd.DataFrame({
        "match_id": int(match_id),
        "event_id": events_df['id'],
        "index": events_df.index,  # Keep original event ID for traceability
        "ingested_at": pd.Timestamp.now(tz="UTC"),
        # Stringify each row dictionary into a raw JSON string
        "event_payload": [
            json.dumps({k: v for k, v in record.items() if k not in ['id', 'index', 'match_id']})
            for record in json_records
        ]
    })
    
    # 4. Stream to BigQuery. 
    # Use append=False on the VERY FIRST iteration to clear old run tracking, then True.
    is_append = False if i == 1 else True
    pipeline.upload_dataframe(staging_df, GCP_DATASET, "sb_events_raw", append=is_append)
    
    time.sleep(0.2)

print("✅ Success! Raw telemetry securely isolated inside BigQuery.")

print("🚀 All events streamed successfully with zero missing columns!")
```
